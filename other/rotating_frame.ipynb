# Rotating reference frames

Kinematics of a non-inertial frame that may **rotate *and* translate**, built from the single pose transform $\mathbf p^A = \mathbf R^A_B\,\mathbf p^B + \mathbf t^A$ and differentiated from first principles.

**Setup.** $\mathbf p^B,\mathbf v^B,\mathbf a^B$ are the *particle of interest* in the body frame $B$; $\mathbf R^A_B,\mathbf t^A$ are the *pose* of $B$ in $A$, with rates $\boldsymbol\omega^B_{AB}$ (spin) and $\dot{\mathbf t}^A$ (origin velocity).

**Contents**
- **Velocity & acceleration** ($B\to A$) — the *transport theorem*: differentiate the pose transform once for $\mathbf v^A$, again for $\mathbf a^A$:
$$\mathbf v^A = \dot{\mathbf t}^A + \mathbf R^A_B\big(\boldsymbol\omega^B_{AB}\times\mathbf p^B + \mathbf v^B\big),$$
$$\mathbf a^A = \ddot{\mathbf t}^A + \mathbf R^A_B\big[\,\underbrace{\mathbf a^B}_{\text{relative}} + \underbrace{2\,\boldsymbol\omega^B_{AB}\times\mathbf v^B}_{\text{Coriolis}} + \underbrace{\boldsymbol\omega^B_{AB}\times(\boldsymbol\omega^B_{AB}\times\mathbf p^B)}_{\text{centripetal}} + \underbrace{\dot{\boldsymbol\omega}^B_{AB}\times\mathbf p^B}_{\text{Euler}}\,\big].$$
The moving origin $\ddot{\mathbf t}^A$ rides outside; the four body-frame terms sit inside $\mathbf R^A_B[\cdots]$. This single-frame $\mathbf a^A$ is the **kick-start** reused at every layer of the nested stack, and is applied numerically in the worked example.
- **Nested transforms** ($C\to B\to A$) — compose the two poses and differentiate. The inner particle rides on a **total-spin transport** $\boldsymbol\omega^C_{AC}=\boldsymbol\omega^C_{AB}+\boldsymbol\omega^C_{BC}$ (cross-Coriolis / gyroscopic couplings fold inside $\boldsymbol\omega^C_{AC}$ and its rate), while B's own spin sweeps the **mount offset** $\mathbf t^B$. The acceleration comes out as **two parallel four-term brackets** — one on the particle $\mathbf p^C$ (total spin $\boldsymbol\omega^C_{AC}$), one on the offset $\mathbf t^B$ (B's spin $\boldsymbol\omega^B_{AB}$) — plus the outer origin $\ddot{\mathbf t}^A$. This is **the general case**; a static mount, fixed origins, or coincident origins are just specializations that zero terms — coincident origins tying to $SE(3)/SE_2(3)$ composition.
- **Worked example** — one horizontal rotating frame read two ways, a **carousel** and a **turning car**, with numbers: a steady turn *is* a carousel, differing only in where you put the origin.

**Conventions:** superscript = destination frame ($\mathbf R^A_B$ maps B→A); $[\cdot]_\times$ = skew/hat operator; $\boldsymbol\omega_{AB}$ = angular velocity of B relative to A, from $\dot{\mathbf R}^A_B = \mathbf R^A_B[\boldsymbol\omega^B_{AB}]_\times = [\boldsymbol\omega^A_{AB}]_\times\,\mathbf R^A_B$.

# Transform ($B\to A$)

## Velocity

$$\begin{aligned}
\mathbf p^A       & = \mathbf t^A + \mathbf R^A_B\, \mathbf p^B \\
\dot{\mathbf p}^A & = \dot{\mathbf t}^A + \dot{\mathbf R}^A_B\, \mathbf p^B + \mathbf R^A_B\, \dot{\mathbf p}^B \\
                  & = \dot{\mathbf t}^A + \mathbf R^A_B\,(\boldsymbol\omega^B_{AB} \times \mathbf p^B) + \mathbf R^A_B\, \mathbf v^B \\
\mathbf v^A       & = \dot{\mathbf t}^A + \mathbf R^A_B\big(\underbrace{\boldsymbol\omega^B_{AB} \times \mathbf p^B}_{\text{transport}} + \underbrace{\mathbf v^B}_{\text{relative}}\big)
\end{aligned}$$

## Acceleration

Starting from $\mathbf v^A = \dot{\mathbf t}^A + \mathbf R^A_B\big(\boldsymbol\omega^B_{AB}\times\mathbf p^B + \mathbf v^B\big)$:

$$\begin{aligned}
\mathbf a^A &= \ddot{\mathbf t}^A + \dot{\mathbf R}^A_B\big(\boldsymbol\omega^B_{AB}\times\mathbf p^B + \mathbf v^B\big) + \mathbf R^A_B\big(\dot{\boldsymbol\omega}^B_{AB}\times\mathbf p^B + \boldsymbol\omega^B_{AB}\times\dot{\mathbf p}^B + \dot{\mathbf v}^B\big) \\
            &= \ddot{\mathbf t}^A + \mathbf R^A_B\big(\boldsymbol\omega^B_{AB}\times(\boldsymbol\omega^B_{AB}\times\mathbf p^B + \mathbf v^B)\big) + \mathbf R^A_B\big(\dot{\boldsymbol\omega}^B_{AB}\times\mathbf p^B + \boldsymbol\omega^B_{AB}\times\mathbf v^B + \mathbf a^B\big) \\
            &= \ddot{\mathbf t}^A + \mathbf R^A_B\big[\,\underbrace{\mathbf a^B}_{\text{relative}} + \underbrace{2\,\boldsymbol\omega^B_{AB}\times\mathbf v^B}_{\text{Coriolis}} + \underbrace{\boldsymbol\omega^B_{AB}\times(\boldsymbol\omega^B_{AB}\times\mathbf p^B)}_{\text{centripetal}} + \underbrace{\dot{\boldsymbol\omega}^B_{AB}\times\mathbf p^B}_{\text{Euler}}\,\big]
\end{aligned}$$

# Transform ($C\to B\to A$)

Compose:

$$\begin{aligned}
\mathbf p^A &= \mathbf t^A + \mathbf R^A_B\,\mathbf p^B\\
\mathbf p^B &= \mathbf t^B + \mathbf R^B_C\,\mathbf p^C\\[4pt]
\mathbf p^A &= \mathbf R^A_C\,\mathbf p^C + \big(\mathbf R^A_B\mathbf t^B + \mathbf t^A\big) \\
\mathbf R^A_C &\equiv \mathbf R^A_B\mathbf R^B_C
\end{aligned}$$

## Velocity

The single-level result applies at **each layer** (keep every $\dot{\mathbf t}$):
$$\begin{aligned}
\mathbf v^A &= \dot{\mathbf t}^A + \mathbf R^A_B\big(\boldsymbol\omega^B_{AB}\times\mathbf p^B + \mathbf v^B\big) \\
\mathbf v^B &= \dot{\mathbf t}^B + \mathbf R^B_C\big(\boldsymbol\omega^C_{BC}\times\mathbf p^C + \mathbf v^C\big)
\end{aligned}$$

Nest the inner into the outer — factor $\mathbf R^A_B$ out, $\mathbf R^B_C$ inside. Since $\mathbf p^B=\mathbf R^B_C\mathbf p^C+\mathbf t^B$, B's spin $\boldsymbol\omega^B_{AB}$ hits **both** the inner body $\mathbf R^B_C\mathbf p^C$ **and** the offset $\mathbf t^B$:
$$\begin{aligned}
\mathbf v^A &= \dot{\mathbf t}^A + \mathbf R^A_B\Big[\,\dot{\mathbf t}^B + \boldsymbol\omega^B_{AB}\times\mathbf t^B + \mathbf R^B_C\big(\boldsymbol\omega^C_{AB}\times\mathbf p^C\big) + \mathbf R^B_C\big(\boldsymbol\omega^C_{BC}\times\mathbf p^C + \mathbf v^C\big)\,\Big] \\
    &= \dot{\mathbf t}^A + \mathbf R^A_B\Big[\,\dot{\mathbf t}^B + \boldsymbol\omega^B_{AB}\times\mathbf t^B + \mathbf R^B_C\big(\boldsymbol\omega^C_{AB}\times\mathbf p^C + \boldsymbol\omega^C_{BC}\times\mathbf p^C + \mathbf v^C\big)\,\Big] \\
    &= \dot{\mathbf t}^A + \mathbf R^A_B\Big[\,\dot{\mathbf t}^B + \boldsymbol\omega^B_{AB}\times\mathbf t^B + \mathbf R^B_C\big((\underbrace{\boldsymbol\omega^C_{AB}}_{B\text{ transport}} + \underbrace{\boldsymbol\omega^C_{BC}}_{C\text{ transport}})\times\mathbf p^C + \mathbf v^C\big)\,\Big] \\
    &= \underbrace{\dot{\mathbf t}^A}_{B\text{ origin}} + \mathbf R^A_B\Big[\,\underbrace{\dot{\mathbf t}^B}_{C\text{ origin}} + \underbrace{\boldsymbol\omega^B_{AB}\times\mathbf t^B}_{\text{offset transport}} + \mathbf R^B_C\big(\underbrace{\boldsymbol\omega^C_{AC}\times\mathbf p^C}_{\text{total transport}} + \underbrace{\mathbf v^C}_{\text{relative}}\big)\,\Big] \\[4pt]
\boldsymbol\omega^C_{AC} &\equiv \boldsymbol\omega^C_{AB} + \boldsymbol\omega^C_{BC}
\end{aligned}$$

## Acceleration

No need to recurse — flatten the composed position and differentiate twice. With $\mathbf R^A_C=\mathbf R^A_B\mathbf R^B_C$ and $\mathbf u^C\equiv\boldsymbol\omega^C_{AC}\times\mathbf p^C+\mathbf v^C$:
$$\mathbf p^A = \mathbf t^A + \mathbf R^A_B\,\mathbf t^B + \mathbf R^A_C\,\mathbf p^C,\qquad
\mathbf v^A = \dot{\mathbf t}^A + \tfrac{d}{dt}\!\big(\mathbf R^A_B\mathbf t^B\big) + \mathbf R^A_C\,\mathbf u^C.$$

Each affine block $\mathbf R\,\mathbf x$ differentiates to the same four-term bracket ($\dot{\mathbf R}^A_B=\mathbf R^A_B[\boldsymbol\omega^B_{AB}]_\times$, $\dot{\mathbf R}^A_C=\mathbf R^A_C[\boldsymbol\omega^C_{AC}]_\times$, $\dot{\mathbf p}^C=\mathbf v^C$, $\dot{\mathbf v}^C=\mathbf a^C$):
$$\begin{aligned}
\mathbf a^A &= \ddot{\mathbf t}^A + \underbrace{\mathbf R^A_B\big[\,\ddot{\mathbf t}^B + 2\,\boldsymbol\omega^B_{AB}\times\dot{\mathbf t}^B + \boldsymbol\omega^B_{AB}\times(\boldsymbol\omega^B_{AB}\times\mathbf t^B) + \dot{\boldsymbol\omega}^B_{AB}\times\mathbf t^B\,\big]}_{\text{same four terms, on the offset }\mathbf t^B} \\
    &\qquad\quad + \mathbf R^A_C\big[\,\underbrace{\mathbf a^C}_{\text{relative}} + \underbrace{2\,\boldsymbol\omega^C_{AC}\times\mathbf v^C}_{\text{Coriolis}} + \underbrace{\boldsymbol\omega^C_{AC}\times(\boldsymbol\omega^C_{AC}\times\mathbf p^C)}_{\text{centripetal}} + \underbrace{\dot{\boldsymbol\omega}^C_{AC}\times\mathbf p^C}_{\text{Euler}}\,\big] \\[4pt]
\boldsymbol\omega^C_{AC} &\equiv \boldsymbol\omega^C_{AB} + \boldsymbol\omega^C_{BC} \\
\dot{\boldsymbol\omega}^C_{AB} &= \boldsymbol\omega^C_{AB}\times\boldsymbol\omega^C_{BC} + \mathbf R^C_B\,\dot{\boldsymbol\omega}^B_{AB}
\end{aligned}$$

# Worked example — carousel & turning car

One horizontal rotating frame, read two ways. The spin/turn axis is vertical, $\boldsymbol\omega=\omega\,\hat z$ (the carousel's spin rate, or the car's **yaw rate** $\dot\psi$). From the body-frame form, an object at body position $\mathbf p^B$ moving at $\mathbf v^B$ with relative accel $\mathbf a^B$ has

$$\mathbf a^A=\ddot{\mathbf t}^A+\mathbf R^A_B\big[\underbrace{\mathbf a^B}_{\text{relative}}+\underbrace{2\boldsymbol\omega\times\mathbf v^B}_{\text{Coriolis}}+\underbrace{\boldsymbol\omega\times(\boldsymbol\omega\times\mathbf p^B)}_{\text{centripetal}}+\underbrace{\dot{\boldsymbol\omega}\times\mathbf p^B}_{\text{Euler}}\big].$$

Each bracket term is a **body-frame** acceleration — what you experience in the rotating frame; $\mathbf R^A_B$ merely rotates their sum into A. So we label them $\mathbf a^B_{\text{centripetal}}$, $\mathbf a^B_{\text{Coriolis}}$, etc. The leading $\ddot{\mathbf t}^A$ is the frame's own **moving-origin** acceleration, sitting *outside* the bracket — zero when the origin is fixed (carousel), nonzero when it rides along (car). (Take $g\approx10\ \mathrm{m/s^2}$ just to gauge magnitudes in "$g$"s.)

## Carousel (origin on the axis)

Body origin at the **center** (on the spin axis), so $\ddot{\mathbf t}^A=\mathbf 0$. Body axes: $\hat x$ = **radial** (outward), $\hat y$ = **tangential**, $\hat z$ = **up** — coinciding with the spin axis $\boldsymbol\omega$. Constant spin ($\dot{\boldsymbol\omega}=0$). Numbers: $\omega=1\ \mathrm{rad/s}$ ($\approx$ 9.5 rpm), radius $r=2$ m.

**Sitting still** — $\mathbf p^B=(r,0,0)$, $\mathbf v^B=\mathbf a^B=\mathbf 0$:
$$\mathbf a^B_{\text{centripetal}}=\boldsymbol\omega\times(\boldsymbol\omega\times\mathbf p^B)=-\omega^2 r\,\hat x\quad(\text{inward, along }-\hat x,\ \ \omega^2 r=2\ \mathrm{m/s^2}\approx0.2g).$$
Inertially you circle at $\boldsymbol\omega\times\mathbf p=\omega r=2$ m/s (tangential, $\hat y$).

**Walking outward** — $\mathbf v^B=(u,0,0)$, $u=1$ m/s:
$$\mathbf a^B_{\text{Coriolis}}=2\boldsymbol\omega\times\mathbf v^B=2\omega u\,\hat y\quad(\text{tangential, along }\hat y,\ \ 2\omega u=2\ \mathrm{m/s^2}).$$
A $2\ \mathrm{m/s^2}$ tangential deflection while you walk; stop and it vanishes, the centripetal stays.

## Turning car (yaw rate $\omega=v/R$)

A car rounds a curve of radius $R$ at speed $v$; its body frame is $\hat x$ = **forward**, $\hat y$ = **left**, $\hat z$ = **up** (the yaw axis, coinciding with $\boldsymbol\omega$). Note the axes are rotated vs the carousel: here **forward $\hat x$ is tangential**, and **left $\hat y$ is radial** (pointing toward the center on a left turn). It **yaws** at $\omega=\dot\psi=v/R$. Numbers: $v=20$ m/s (72 km/h), $R=100$ m $\Rightarrow\ \omega=0.2\ \mathrm{rad/s}\approx11.5°/\mathrm{s}$.

The body **origin rides with the car** (off the curve's center, radius $R$), so the car's circular motion lives in the moving origin. For a point **at the car origin** ($\mathbf p^B=\mathbf v^B=\mathbf a^B=\mathbf 0$) the whole bracket vanishes, and the centripetal you feel is entirely $\ddot{\mathbf t}^A$:
$$\ddot{\mathbf t}^A=\frac{v^2}{R}\,\hat y\quad(\text{lateral, toward the turn center},\ \ v^2/R=4\ \mathrm{m/s^2}\approx0.4g).$$

## Same physics, two rides

A steady turn **is** a carousel. The car's lateral $v^2/R$ equals the carousel's $\omega^2 R$ with $R=v/\omega$ — both are just

$$\boxed{\ a_{\text{centripetal}}=\omega^2\times(\text{distance from the turn center}),\ \text{pointing toward center}.\ }$$

The only difference is bookkeeping — **where you put the origin**:
- **Carousel:** origin **on the axis**, so the centripetal appears as the body term $\boldsymbol\omega\times(\boldsymbol\omega\times\mathbf p^B)$ with $\mathbf p^B=r\,\hat x$ (and $\ddot{\mathbf t}^A=\mathbf 0$).
- **Car:** origin **on the vehicle** (off the center), so the same centripetal appears via the moving origin $\ddot{\mathbf t}^A$. An offset passenger (back seat, $\mathbf p^B\neq\mathbf 0$) picks up an *extra* body-term $\boldsymbol\omega\times(\boldsymbol\omega\times\mathbf p^B)$, and if the driver changes the turn rate, an **Euler** $\dot{\boldsymbol\omega}\times\mathbf p^B$ shove.
